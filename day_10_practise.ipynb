We will not do all this steps always,

We create a pipeline and pass data through it.

```python
### Creating a Data Analysis pipeline

def load_path(path):
  return df.read.csv(path)


def clean_data(path):
  df=df.dropna(subset=["amount"])
  df["date"] = pd.to_datetime(df["date"])
  return df


def transform_data(df):
  df["net_amount"] = df["amount"] - df["discount"]
  return df


def run_pipeline(path):
  return transform_data(
    clean_data(
      load_data(path)
    )
  )


run_pipeline("transactions.csv")
```
